# **Import libraries**

```python
import torch

from torchvision import transforms as T

import tkinter as tk
from tkinter import filedialog

from PIL import ImageTk, Image
import glob
```

# **Load model**

```python
model = torch.load('foodModel.pt', map_location=torch.device('cpu'))
```

# **To device**

```python
if torch.cuda.is_available() == True:
    device = 'cuda'
else:
    device = 'cpu'
```

# **Graphics**

```python
root = tk.Tk()
root.title("Food detector")

width = 625
height = 525

sc_width = root.winfo_screenwidth()
sc_height = root.winfo_screenheight()

x = (sc_width - width) // 2
y = (sc_height - height) // 2

root.geometry(f"{width}x{height}+{x}+{y}")
root.resizable(False, False)
root.minsize(625,525)
root.config(bg = '#7ea8a8')
```

```python
f1 = tk.Frame(root,bg='#7ea8a8')
finfo = tk.Frame(root,bg='#7ea8a8')
fexit = tk.Frame(root,bg='#7ea8a8')
```

```python
def info():
    finfo.tkraise()

    l1 = tk.Label(finfo,text='',font=('Franklin Gothic Medium',11),bg='#7ea8a8')
    l1.config(text="""USER GUIDE:
    This app is created to detect 21 different types of foods. It is powered by an AI deep-learning
    CNN model using Pytorch. It has been trained based on a wide range of foods
    and is 80.4 percent accurate in its detection.

    Send an RGB picture with any resolution using any camera to your PC.
        
    Place a shortcut of the "Food Detector" application on your desktop or
    open it directly from the original folder.
        
    Note: Make sure that Python and Pytorch are installed on your computer.
    Note:Make sure FoodGraphics.ipynb - FoodModel.ipynb - foodModel.pt are available
        
    Hope you enjoy the app ;)
        
    The developer, Matin.B""")
    
    b1 = tk.Button(finfo,text='Back to main menu',font=('Franklin Gothic Medium',24),bg='#a89c92',command=frame_1)
    
    l1.grid(row=0,column=0,padx=5,pady=5,sticky="NEWS")
    b1.grid(row=1,column=0,padx=5,pady=5,sticky="NEWS")

    for i in range(2):
        for j in range(1):
            finfo.grid_rowconfigure(i,weight=1)
            finfo.grid_columnconfigure(j,weight=1)
```

```python
def frame_exit():
    fexit.tkraise()

    l1 = tk.Label(fexit,text='Are you sure that you want to exit?',font=('Franklin Gothic Medium',22),bg='#7ea8a8')
    
    def exit_tk():
        root.destroy()

    b1 = tk.Button(fexit,text='Yes',font=('Franklin Gothic Medium',22),bg='#a89c92',command=exit_tk)
    b2 = tk.Button(fexit,text='No',font=('Franklin Gothic Medium',22),bg='#a89c92',command=frame_1)    

    l1.grid(row=0,column=0,columnspan=2,rowspan=2,padx=5,pady=5,sticky="NEWS")
    b1.grid(row=2,column=0,padx=5,pady=5,sticky="NEWS")
    b2.grid(row=2,column=1,padx=5,pady=5,sticky="NEWS")

    for i in range(3):
        for j in range(2):
            fexit.grid_rowconfigure(i,weight=1)
            fexit.grid_columnconfigure(j,weight=1)
```

```python
def frame_1():
    f1.tkraise()
    
    l1 = tk.Label(f1,text='Food Detector',font=('Franklin Gothic Medium',28),bg='#7ea8a8')
    l2 = tk.Label(f1,text='Choose an image to detect!',font=('Franklin Gothic Medium',20),bg='#7ea8a8')
    l3 = tk.Label(f1,text='',font=('Franklin Gothic Medium',22),bg='#7ea8a8')

    def detect():
        try:
            img_path = filedialog.askopenfilename(filetypes=[('Image Files', ('*.png;*.jpg;*.jpeg'))])
            
            pic = Image.open(img_path)
            transform = T.Compose([T.Resize((128,128)),
                                   T.ToTensor(),
                                   T.Normalize(mean=[0.485, 0.456, 0.406],
                                               std=[0.229, 0.224, 0.225])])
            pic_tensor = transform(pic)
            pic_cuda = pic_tensor.to(device)
            output = model(pic_cuda.unsqueeze(0))
            toname={0: 'baklava',
             1: 'Caesar salad',
             2: 'Cheesecake',
             3: 'Cheese sandwich',
             4: 'Chicken curry',
             5: 'Chicken wings',
             6: 'Chocolate cake',
             7: 'Donuts',
             8: 'Eggs',
             9: 'Falafel',
             10: 'Fish',
             11: 'French fries',
             12: 'Hamburger',
             13: 'Hotdog',
             14: 'Ice cream',
             15: 'Lasagna',
             16: 'Omelette',
             17: 'Pizza',
             18: 'Spaghetti',
             19: 'Steak',
             20: 'Sushi'}
            result = toname[output.argmax().to('cpu').item()]
            l3.config(text='Detecting...')
            def update_label():
                l3.config(text=f"{result} has been detected!")
            root.after(3000,update_label)
        except AttributeError:
            l3.config(text="No image has been chosen yet!")
    
    b1 = tk.Button(f1,text='🛈',font=('Franklin Gothic Medium',35),bg='#a89c92',command=info)
    b2 = tk.Button(f1,text='Exit',font=('Franklin Gothic Medium',18),bg='#a89c92',command=frame_exit)
    b3 = tk.Button(f1,text='Choose an image',font=('Franklin Gothic Medium',18),bg='#a89c92',command=detect)

    l1.grid(row=0,column=0,columnspan=3,padx=5,pady=5,sticky="NEWS")
    l2.grid(row=1,column=0,columnspan=3,padx=5,pady=5,sticky="NEWS")
    l3.grid(row=4,column=0,columnspan=3,padx=5,pady=5,sticky="NEWS")

    b1.grid(row=0,column=2,padx=5,pady=5,sticky="NE")
    b3.grid(row=2,column=0,columnspan=2,rowspan=2,padx=5,pady=5,sticky="NEWS")
    b2.grid(row=2,column=2,rowspan=2,padx=5,pady=5,sticky="NEWS")

    for i in range(5):
        for j in range(3):
            f1.grid_rowconfigure(i,weight=1)
            f1.grid_columnconfigure(j,weight=1)
```

```python
frame_1()
f1.grid(row=0,column=0,sticky="NEWS")
fexit.grid(row=0,column=0,sticky="NEWS")
finfo.grid(row=0,column=0,sticky="NEWS")
root.grid_rowconfigure(0,weight=1)
root.grid_columnconfigure(0,weight=1)
```

```python
tk.mainloop()
```
